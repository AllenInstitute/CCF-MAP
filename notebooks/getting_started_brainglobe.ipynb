# Getting started 
Let us start by selecting the Allen mouse CCFv3 as a test case. All data are stored on a public S3 bucket. 

All atlases have four major components: the **template**, **coordinate space**, **annotation**, and **terminology**

1. The **template** is a reference image volume that represents the contrast features of the anatomical region. In the macaque case, the anatomical template is the Mac25Rhesus template, which is comprised of an averaged T1 weighted MRI from 25 adult rhesus macaques and resampled to 160 um voxel resolution. 

2. The template also defines the **coordinate space**. The coordinate space is the mathematical representation of the anatomical space and defines the orientation and direction of the x,y,z axes. All datasets aligned to the same coordinate space can be visualized together. Datasets in different coordinate spaces need to be registered using diffeomorphic registration tools. 

3. The **annotation** is a set of manually defined segmentations of the template space. These segmentations parcellate the template into defined structures. Each segmentation is assigned a unique numeric value.   

4. Annotations are named and labeled using the **Harmonized Ontology of Mammalian Brain Anatomy (HOMBA)**, which provides a **terminology**, a taxonomy of the structures and their relationship along with relevant metadata (eg. label colors), and serves as a lookup between the annotation value and the label.

Together, these components uniquely define an anatomical atlas
